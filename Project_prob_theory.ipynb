Задача 1.8

Доказательство: Теорема Кантора-Бернштейна гласит, что если мощность множества A не превосходит мощность множества B, а мощность множества B не превосходит мощность множества A, то эти множества равномощны. При этом мы знаем, что если существует инъекция из множества А во множество В, то по определению инъекции мощность множества А не больше мощности множества В. Значит, чтобы применить теорему и доказать задачу, достаточно привести инъекцию из отрезка в интервал и наоборот

Инъекция из (0,1) в [0,1]: f(x) = x. $\forall x \in (0,1): f(x) = x \in [0,1]$.
При этом то, что это инъекция, очевидно: если x $\neq$ y, то f(x) = x $\neq$ y = f(y)

Инъекция из [0,1] в (0,1): g(x) = 0.5x + 0.25. Это линейная функция, при этом $g(0) = 0.5 * 0 + 0.25 = 0.25$ и $g(1) = 0.5 * 1 + 0.25 = 0.75$, то есть $\forall x \in [0,1]: g(x) = x \in [0.25,0.75] \subset (0,1)$
Инъекция также очевидна в силу линейности: $g(x) = g(y) ⇔ 0.5x + 0.25 = 0.5y + 0.25 ⇔ 0.5x = 0.5y ⇔ x = y$

Значит, $|(0,1)| \leq |[0,1]|$ и $|[0,1]| \leq |(0,1)|$, а значит по теореме Кантора Бернштейна данные отрезок и интервал равномощны. Что и требовалось доказать

Задача 1.20

Решение:

С одной стороны мы можем просто взять k человек из n без учета порядка, это будет $C_n^k$.

С другой стороны, рассмотрим разбиение событий на 2: Алексей попал в группу и Алексей не попал в группу. Посчитаем количество вариантов реализации каждого из этих двух событий. Если Алексей попал в группу, то нам осталось набрать k - 1 человека из n - 1 оставшихся, а это можно сделать $C_{n - 1}^{k - 1}$ вариантами. Если же Алексей не попал в группу, то нам до сих пор нужно выбрать k человек, но уже из n - 1 (так как Алексея мы убрали). Получается количество способов $C_{n - 1}^k$. А суммарно они составляют изначальное событие по выбору k человек из n, вот мы и получаем формулу Паскаля: $C_n^k = C_{n - 1}^{k - 1} + C_{n - 1}^k$

P.S. формулу Паскаля ещё несложно доказать в лоб через факториалы, но этого в задаче вроде не требуется, да и комбинаторное решение гораздо красивее

Задача 1.22

Решение:

Пусть $X$ и $Y$ — время прибытия первого и второго студента соответственно, выраженное в минутах после 9:00.
Так как студенты приходят в течение одного часа, то их времена прихода ограничены пределами:
$$0 \le X \le 60, \quad 0 \le Y \le 60$$

Множество всех возможных исходов представляет собой квадрат $\Omega$ со стороной 60. Его площадь равна:
$$S(\Omega) = 60 \times 60 = 3600$$

Студенты встретятся, если разница во времени их прихода не превысит 15 минут. Это условие задает множество благоприятных исходов $A$, описываемое неравенством:
$$|X - Y| \le 15 \iff -15 \le X - Y \le 15$$

Раскрывая модуль, получаем систему ограничений для области встречи:
$$Y \le X + 15 \quad \text{и} \quad Y \ge X - 15$$

Площадь благоприятной области $S(A)$ проще всего найти, вычтя из площади всего квадрата площади двух неблагоприятных треугольников в углах (каждый из которых имеет катеты длиной $60 - 15 = 45$):
$$S(A) = S(\Omega) - 2 \times \left(\frac{1}{2} \times 45 \times 45\right) = 3600 - 45^2 = 3600 - 2025 = 1575$$

Искомая вероятность встречи равна отношению площадей:
$$P(A) = \frac{S(A)}{S(\Omega)} = \frac{1575}{3600} = \frac{7}{16} = 0.4375$$

 события независимые, значит вероятностью общего события это произведение

1.   События независимые, значит вероятностью общего события это произведение вероятностей исходов: $P(A \cap B) = P(A) * P(B)$. У нас 5 правильных классификаций и 1 неправильная, распределение биномиальное, значит итоговая вероятность будет равна $P(X = 5) = C_6^5 * 0.8^5 * 0.2 = 0.393216$
2.   Тут пойдем через отрицание: P(модель допустит хотя бы 1 ошибку) = 1 - P(модель не допустит ни одну ошибку) =  1 - Р(все 6 классификаций верные). А вероятность всех верных классификаций мы считаем по формуле биномиального распределения и вероятности независимых событий аналогично пункту 1: P(все 6 классификаций верные) = $0.8^6 = 0.262144 ⇒$ P(модель допустит хотя бы 1 ошибку) = $1 - 0.262144 = 0.737856$
3. Тут воспользуемся формулой условной вероятности: Р(5 верных | хотя-бы 1 ошибка) = Р(5 верных и хотя бы 1 ошибка) / Р(хотя бы 1 ошибка). Вероятность знаменателя мы уже посчитали в пункте 2, это 0.737856, а в числителе легко заметить, что если у нас ровно 5 верных предсказаний, то значит ровно 1 ошибка и второе условие выполняется автоматически, то есть P(5 верных и хотя бы 1 ошибка) = Р(5 верных) = 0.393216 из пункта 1. Делим и получаем ответ: 0.532917

Задача 1.25

Решение:
По условию задачи число звонков это распределение Пуассона. Формула Пуассона определяет вероятность получить ровно $k$ событий за фиксированный интервал времени:
$P(X = k) = \frac{\lambda^k \cdot e^{-\lambda}}{k!}$

Так как мы знаем, что за 8 часов в среднем поступает 16 звонков, значит пересчитаем интенсивность для 1 часа и получим, что за 1 час в среднем будет поступать два звонка, то есть $\lambda = 2$

Теперь нам необходимо найти вероятность события $P(2 \le X \le 4)$. Это сумма вероятностей того, что за час поступит ровно 2, 3 или 4 звонка:
$$P(2 \le X \le 4) = P(X = 2) + P(X = 3) + P(X = 4)$$

Посчитаем каждое слагаемое отдельно при $\lambda = 2$:

* **Для $k = 2$:**
  $$P(X = 2) = \frac{2^2 \cdot e^{-2}}{2!} = \frac{4 \cdot e^{-2}}{2} = 2 \cdot e^{-2} \approx 2 \cdot 0.135335 = 0.270671$$

* **Для $k = 3$:**
  $$P(X = 3) = \frac{2^3 \cdot e^{-2}}{3!} = \frac{8 \cdot e^{-2}}{6} = \frac{4}{3} \cdot e^{-2} \approx 1.333333 \cdot 0.135335 = 0.180447$$

* **Для $k = 4$:**
  $$P(X = 4) = \frac{2^4 \cdot e^{-2}}{4!} = \frac{16 \cdot e^{-2}}{24} = \frac{2}{3} \cdot e^{-2} \approx 0.666667 \cdot 0.135335 = 0.090224$$

Складываем полученные значения и получаем ответ:
$P(2 \le X \le 4) \approx 0.270670 + 0.180447 + 0.090223 = 0.541340 \approx 0.541$
